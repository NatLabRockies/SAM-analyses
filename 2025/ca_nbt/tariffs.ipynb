```python
import pandas as pd
import numpy as np
import sklearn
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Current Year NBT Pricing Upload MIDAS.csv')
```

```python
prices = df.Value.values
```

```python
price_2025 = np.array(prices[0:8760])
price_2026 = np.array(prices[8760:(8760) * 2])
price_2027 = np.array(prices[(8760) * 2:(8760) * 3])
leap_day_total = (8760) * 4 + 24
price_2028 = np.array(prices[(8760) * 3:leap_day_total])
price_2029 = np.array(prices[leap_day_total:leap_day_total + 8760])
price_2030 = np.array(prices[leap_day_total+ 8760:leap_day_total + (8760) * 2])
price_2031 = np.array(prices[leap_day_total+ (8760) * 2:leap_day_total + (8760) * 3])
```

```python
print("2026 MAPE", sklearn.metrics.mean_absolute_percentage_error(price_2026, price_2025))
```

```python
print(max(price_2026 - price_2025))
```

```python
fig = plt.figure(figsize=(18, 6))
ax1 = fig.add_subplot(111)
ax1.plot((price_2025)[5000:5400], label="2025")
ax1.plot((price_2026)[5000:5400], label="2026")
ax1.plot((price_2027)[5000:5400], label="2027")

handles, labels = ax1.get_legend_handles_labels()
ax1.legend(handles, labels)
```
